```python
# Import libraries required for regression

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
# Select independent variables (X)
X = df[['Unit_Price', 'Discount_Applied', 'Customer_Age']]

# Select dependent variable (Y)
y = df['Units_Sold']
```

```python
# Split data into training and testing sets
# 80% for training and 20% for testing

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training records:", len(X_train))
print("Testing records:", len(X_test))
```

```python
# Create Linear Regression model

model = LinearRegression()

# Train the model
model.fit(X_train, y_train)

print("Regression model trained successfully!")
```

```python
# Predict Units Sold using the test data

y_pred = model.predict(X_test)

print("Predicted Units Sold:")
print(y_pred[:10])
```

```python
# Calculate regression performance metrics

r2 = r2_score(y_test, y_pred)
mae = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))

print("----- Regression Results -----")
print("R² Score:", round(r2, 4))
print("Mean Absolute Error (MAE):", round(mae, 2))
print("Root Mean Squared Error (RMSE):", round(rmse, 2))
```

```python
# Display the effect of each independent variable

print("----- Regression Coefficients -----")

for feature, coefficient in zip(X.columns, model.coef_):
    print(feature, ":", round(coefficient, 4))

print("Intercept:", round(model.intercept_, 4))
```

```python
# Confidence Interval Analysis

from scipy import stats

# Use the random sample of 100 transactions
sales_sample = sample['Final_Sales']

# Calculate sample mean
sample_mean = sales_sample.mean()

# Calculate sample standard deviation
sample_std = sales_sample.std()

# Sample size
n = len(sales_sample)

# Calculate standard error
standard_error = sample_std / np.sqrt(n)

# Calculate 95% confidence interval
confidence_level = 0.95
degrees_of_freedom = n - 1

t_value = stats.t.ppf(
    (1 + confidence_level) / 2,
    degrees_of_freedom
)

margin_of_error = t_value * standard_error

lower_limit = sample_mean - margin_of_error
upper_limit = sample_mean + margin_of_error

print("----- 95% Confidence Interval -----")
print("Sample Mean Sales: ₹", round(sample_mean, 2))
print("Margin of Error: ₹", round(margin_of_error, 2))
print("Lower Limit: ₹", round(lower_limit, 2))
print("Upper Limit: ₹", round(upper_limit, 2))
```

```python
# Visualize the confidence interval

plt.figure(figsize=(10, 3))

plt.errorbar(
    sample_mean,
    0,
    xerr=margin_of_error,
    fmt='o',
    color='steelblue',
    ecolor='black',
    capsize=8,
    markersize=8
)

plt.axvline(
    sample_mean,
    color='red',
    linestyle='--',
    label='Sample Mean'
)

plt.xlabel("Average Final Sales (₹)")
plt.yticks([])
plt.title("95% Confidence Interval for Average Final Sales")
plt.legend()

plt.tight_layout()
plt.show()
```

```python
# Bootstrapping Analysis

# Set random seed for reproducible results
np.random.seed(42)

# Number of bootstrap simulations
n_bootstrap = 1000

# Store bootstrap mean values
bootstrap_means = []

# Perform bootstrap sampling
for i in range(n_bootstrap):
    bootstrap_sample = np.random.choice(
        df['Final_Sales'],
        size=len(df),
        replace=True
    )

    bootstrap_means.append(bootstrap_sample.mean())

# Convert results into a Series
bootstrap_means = pd.Series(bootstrap_means)

print("Number of Bootstrap Samples:", n_bootstrap)
print("Bootstrap Mean Sales: ₹", round(bootstrap_means.mean(), 2))
```

```python
# Calculate 95% Bootstrap Confidence Interval

lower_bootstrap = bootstrap_means.quantile(0.025)
upper_bootstrap = bootstrap_means.quantile(0.975)

print("----- Bootstrap Confidence Interval -----")
print("Lower Limit: ₹", round(lower_bootstrap, 2))
print("Upper Limit: ₹", round(upper_bootstrap, 2))
```

```python
# Hypothesis Testing

from scipy.stats import ttest_1samp

# Set the hypothesized average sales value
hypothesized_mean = 1200

# Perform one-sample t-test
t_statistic, p_value = ttest_1samp(
    df['Final_Sales'],
    hypothesized_mean
)

print("----- One-Sample T-Test -----")
print("Hypothesized Mean: ₹", hypothesized_mean)
print("T-Statistic:", round(t_statistic, 4))
print("P-Value:", round(p_value, 4))
```

```python
# Set significance level
alpha = 0.05

if p_value < alpha:
    print("Decision: Reject the Null Hypothesis")
    print("Conclusion: The average Final Sales is significantly different from ₹1200.")
else:
    print("Decision: Fail to Reject the Null Hypothesis")
    print("Conclusion: There is not enough evidence to say that the average Final Sales is different from ₹1200.")
```

```python
# Optimization Modeling

from scipy.optimize import linprog

# Profit coefficients
# linprog minimizes, so use negative profit
profit = [-40, -30]

# Constraints
# 2A + B <= 100
# A + 2B <= 80
constraints = [
    [2, 1],
    [1, 2]
]

# Available resources
resources = [100, 80]

# Product quantities cannot be negative
bounds = [
    (0, None),   # Product A
    (0, None)    # Product B
]

# Solve the optimization problem
result = linprog(
    profit,
    A_ub=constraints,
    b_ub=resources,
    bounds=bounds,
    method='highs'
)

print("----- Optimization Results -----")

if result.success:
    product_A = result.x[0]
    product_B = result.x[1]
    maximum_profit = -result.fun

    print("Optimal Quantity of Product A:", round(product_A, 2))
    print("Optimal Quantity of Product B:", round(product_B, 2))
    print("Maximum Profit: ₹", round(maximum_profit, 2))
else:
    print("Optimization could not be solved.")
```

```python
# Optimization Graph

import numpy as np
import matplotlib.pyplot as plt

A = np.linspace(0, 100, 500)

# Constraints
B_machine = 100 - 2 * A
B_labour = (80 - A) / 2

# Keep values non-negative
B_machine = np.maximum(B_machine, 0)
B_labour = np.maximum(B_labour, 0)

plt.figure(figsize=(10, 6))

# Plot constraints
plt.plot(A, B_machine, label='Machine Constraint')
plt.plot(A, B_labour, label='Labour Constraint')

# Feasible region
B_feasible = np.minimum(B_machine, B_labour)
plt.fill_between(
    A,
    0,
    B_feasible,
    alpha=0.3,
    label='Feasible Region'
)

# Optimal solution
plt.scatter(
    40,
    20,
    color='red',
    s=100,
    edgecolor='black',
    label='Optimal Solution (40, 20)'
)

plt.xlabel("Product A Units")
plt.ylabel("Product B Units")
plt.title("Optimization Model - Feasible Region")
plt.xlim(0, 100)
plt.ylim(0, 100)
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()
```
